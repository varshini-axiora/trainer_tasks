1.Find the second largest element in a list without using sort().



```python
def second_largest(nums):
    largest = second = -float('inf')

    for n in nums:
        if n > largest:
            second = largest
            largest = n
        elif n > second and n != largest:
            second = n

    return second


numbers = [10, 20, 4, 45, 99]
print("Second largest:", second_largest(numbers))
```

2.Find All Pairs with Given Sum. Return all pairs whose sum equals a given target.

```python
def find_pairs(nums, target):
    pairs = []
    
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.append((nums[i], nums[j]))
    return pairs
numbers = [1, 5, 7, -1, 5]
target = 6
print(find_pairs(numbers, target))
```

4.Group words that are anagrams.

Input: ["eat","tea","tan","ate","nat","bat"]

Output: [['eat','tea','ate'], ['tan','nat'], ['bat']]

```python
def group_anagrams(words):
    anagram_dict = {}

    for word in words:
        key = "".join(sorted(word))

        if key not in anagram_dict:
            anagram_dict[key] = []

        anagram_dict[key].append(word)

    return list(anagram_dict.values())


words = ["rat","tar","art","car","arc"]

print(group_anagrams(words))
```

5.Merge two dictionaries. If key exists, add values.
d1 = {'a':10,'b':20}

d2 = {'b':5,'c':15}

Output: {'a':10,'b':25,'c':15}

```python
def merge_dict(d1, d2):
    result = d1.copy()   

    for key in d2:       
        if key in result:
            result[key] = result[key] + d2[key]   
        else:
            result[key] = d2[key]   

    return result


d1 = {'a':25, 'b':60}
d2 = {'b':50, 'c':5}

print(merge_dict(d1, d2))
```

6.Find the longest word words = ["python","machine","AI","data"]
Output: "machine"

```python
def longest_word(words):
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


words = ["varshini", "vinay", "sunithaaa", "raju"]
print("Longest word:", longest_word(words))
```

7.Count number of vowels and consonants in a string.
Input: "python"
Output: Vowels: 1 Consonants: 5

```python
def count_vowels_consonants(text):
    vowels = "aeiouAEIOU"
    v_count = 0
    c_count = 0

    for ch in text:
        if ch.isalpha():       
            if ch in vowels:    
                v_count += 1
            else:               
                c_count += 1

    return v_count, c_count


sentence = "Learning Python is interesting"

v, c = count_vowels_consonants(sentence)

print("Sentence:", sentence)
print("Total Vowels:", v)
print("Total Consonants:", c)
```

9.Compress a String
Input: "aaabbc"

Output: "a3b2c1"

```python
def compress_string(text):
    result = ""
    count = 1

    for i in range(len(text)):

        if i + 1 < len(text) and text[i] == text[i + 1]:
            count += 1
        else:
            result += text[i] + str(count)
            count = 1

    return result

print(compress_string("ppqqqrrr"))
```

10.Generate all permutations of a string.

```python
def permutations_string(s):
    
    if len(s) == 0:
        return [""]

    result = []

    for i in range(len(s)):
        char = s[i]                 
        remaining = s[:i] + s[i+1:] 

        for p in permutations_string(remaining):
            result.append(char + p)

    return result

word = "CAT"
output = permutations_string(word)

for i in output:
    print(i)
```
